```python

import warnings
warnings.filterwarnings("ignore")
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

SEED = 42
np.random.seed(SEED)
sns.set_theme(style="whitegrid", palette="deep")

DATA = Path("/home/workdir/artifacts/Datasets/cleaned")
df = pd.read_parquet(DATA / "survey_2022_cleaned.parquet")
print("2022 cleaned shape:", df.shape)
display(df.head())
display(df.describe(include="all").T.head(15))
```

```python

# Quick demographic snapshot
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df["age"].value_counts().sort_index().plot(kind="bar", ax=axes[0], color="#2A9D8F", edgecolor="white")
axes[0].set_title("Age")
df["gender"].value_counts().plot(kind="bar", ax=axes[1], color="#1B3A4B", edgecolor="white")
axes[1].set_title("Gender")
plt.tight_layout()
plt.show()

print("Top 10 countries:")
display(df["country"].value_counts().head(10))
print("\nRole groups:")
display(df["role_group"].value_counts().head(12))
```
